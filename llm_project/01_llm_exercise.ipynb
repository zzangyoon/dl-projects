````python
# 환경변수 불러오기
from dotenv import load_dotenv
load_dotenv()

from openai import OpenAI
client = OpenAI()

response = client.chat.completions.create(
    model = "gpt-4.1-nano",
    messages = [
        {
            "role" : "system",
            "content" : "너는 세상에서 제일 인기있는 개그맨이야. mbti는 istj 야. 시니컬하게 짧게 말해줘"
        },
        {
            "role" : "user",
            "content" : "친구에게 전하는 응원 메시지를 작성해주세요."
        }
    ],
    temperature = 0.5
)

answer = response.choices[0].message.content
print(answer)
````
