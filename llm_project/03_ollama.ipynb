## 1. LLM 연결

````python
import ollama

# LLM 리스트 조회
local_llm_list = ollama.list()
for llm in local_llm_list:
    print(llm[1])
````

````python
import ollama

client = ollama

def chat_local_llm(user_text):
    system_prompt = """당신은 시니컬한 챗봇입니다."""
    model = "llama3.1:8b"       # 띄어쓰기 오류, 빠름, 
    # model = "gpt-oss:20b"     # 답변이 엄청 오래 걸림, 정확성 높음
    # model = "deepseek-r1:7b"  # 한국어로 질문했는데 답변을 영어로


    response = client.chat(
        model = model,
        messages = [
            {"role" : "system", "content" : system_prompt},
            {"role" : "user", "content" : user_text}
        ]
    )
    
    return response.message.content
````

## 2. 챗봇 만들기

### llama3.1:8b

````python
user_text = "한국어, 영어, 일본어, 중국어, 프랑스어로 자기소개를 해주세요"
answer_llama = chat_local_llm(user_text)
print(answer_llama)
````

### gpt-oss:20b

````python
user_text = "한국어, 영어, 일본어, 중국어, 프랑스어로 자기소개를 간단하게 해주세요"
answer_gpt_oss = chat_local_llm(user_text)
print(answer_gpt_oss)
````

````python
user_text = "한국어, 영어, 일본어, 중국어, 프랑스어로 자기소개를 간단하게 해주세요"
answer_deepseek = chat_local_llm(user_text)
print(answer_deepseek)
````

## 3. 챗봇 대화하기

````python
import ollama

client = ollama

def chat_llm(model, user_text):
    system_prompt = """
    당신은 음식 추천 전문가입니다.
    상대방의 기분, 건강상태, 입맛에 따라 메뉴를 정해주고, 맛집을 서치해주고, 레시피를 알려줄 수 있습니다.
    
    엄마같은 말투로 조언해주듯이 말해주세요.
    """

    response = client.chat(
        model = model,
        messages = [
            {"role" : "system", "content" : system_prompt},
            {"role" : "user", "content" : user_text}
        ]
    )
    
    return response.message.content
````

### llama3.1:8b

````python
while True:
    model = "llama3.1:8b"
    # model = "gpt-oss:20b"
    # model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

### gpt-oss:20b

````python
while True:
    # model = "llama3.1:8b"
    model = "gpt-oss:20b"
    # model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

### deepseek-r1:7b

````python
while True:
    # model = "llama3.1:8b"
    # model = "gpt-oss:20b"
    model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````
