````python
# 페르소나 챗봇 만들기
import ollama

def chat(user_text):
    response = ollama.chat(
        model = "llama3.1:8b",
        messages = [
            {"role" : "system", "content" : "불친절하게 말해주세요"},
            {"role" : "user", "content" : user_text}
        ]
    )

    return response.message.content
````

````python
while True:
    question = input("입력하세요")
    if question == "종료":
        break
    answer = chat(question)
    print(f"Question : {question}")
    print(f"Answer : {answer}")
````

````python
import ollama

client = ollama

def chat_local_llm(user_text):
    system_prompt = """당신은 시니컬한 챗봇입니다."""
    model = "llama3.1:8b"       # 띄어쓰기 오류, 빠름, 
    # model = "gpt-oss:20b"     # 답변이 엄청 오래 걸림, 정확성 높음
    # model = "deepseek-r1:7b"  # 한국어로 질문했는데 답변을 영어로


    response = client.chat(
        model = model,
        messages = [
            {"role" : "system", "content" : system_prompt},
            {"role" : "user", "content" : user_text}
        ]
    )
    
    return response.message.content
````

## 3. 챗봇 대화하기

````python
import ollama

client = ollama

def chat_llm(model, user_text):
    system_prompt = """
    당신은 음식 추천 전문가입니다.
    상대방의 기분, 건강상태, 입맛에 따라 메뉴를 정해주고, 맛집을 서치해주고, 레시피를 알려줄 수 있습니다.
    
    엄마같은 말투로 조언해주듯이 말해주세요.
    """

    response = client.chat(
        model = model,
        messages = [
            {"role" : "system", "content" : system_prompt},
            {"role" : "user", "content" : user_text}
        ]
    )
    
    return response.message.content
````

### llama3.1:8b

````python
while True:
    model = "llama3.1:8b"
    # model = "gpt-oss:20b"
    # model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

### gpt-oss:20b

````python
while True:
    # model = "llama3.1:8b"
    model = "gpt-oss:20b"
    # model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

### deepseek-r1:7b

````python
while True:
    # model = "llama3.1:8b"
    # model = "gpt-oss:20b"
    model = "deepseek-r1:7b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

### gemma3:4b

````python
while True:
    # model = "llama3.1:8b"
    # model = "gpt-oss:20b"
    # model = "deepseek-r1:7b"
    model = "gemma3:4b"

    print("텍스트를 입력하세요")

    user_text = input()
    print(f"입력된 텍스트 : {user_text}")

    answer = chat_llm(model, user_text)
    print(answer)

    if user_text == "종료":
        break
````

# Ollama API로 요청하기

````python
# curl http://localhost:11434/api/generate -d '{
#     "model" : "gemma3:4b",
#     "prompt" : "Why is the sky blue?"
# }'

import requests

url = "http://localhost:11434/api/generate"
data = {
    "model" : "gemma3:4b",
    "prompt" : "Why is the sky blue?"
}
````

````python
response = requests.post(url, json = data)
print(response.text)
````

````python
import json

# generate 는 다 쪼개서 오는데, 이걸 다 합치는 코드
with requests.post(url, json=data) as r:
    for line in r.iter_lines():
        print(json.loads(line)["response"], end="")
````
